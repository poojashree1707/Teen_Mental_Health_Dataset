```python
gender_count = df['gender'].value_counts()

plt.bar(gender_count.index, gender_count.values,color=['blue','pink'])

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python
depression = df['depression_label'].value_counts()

plt.pie(
    depression.values,
    labels=['Not Depressed', 'Depressed'],
    colors=['lavender', 'black'],
    autopct='%1.1f%%'
)

plt.title("Depression Percentage")

plt.show()
```

```python
plt.hist(df['daily_social_media_hours'], bins=10,color='grey')

plt.title("Daily Social Media Usage")
plt.xlabel("Hours")
plt.ylabel("Frequency")

plt.show()
```

```python
platform = df['platform_usage'].value_counts()

colors = ['skyblue', 'royalblue', 'midnightblue']

plt.bar(platform.index, platform.values, color=colors)

plt.title("Platform Usage")
plt.xlabel("Platform")
plt.ylabel("Users")

plt.show()
```

```python
plt.plot(df['age'],
         df['academic_performance'],
         color='purple')

plt.title("Age vs Academic Performance")
plt.xlabel("Age")
plt.ylabel("Academic Performance")

plt.show()
```

```python
plt.scatter(df['daily_social_media_hours'],
            df['sleep_hours'],
            color='blue')

plt.title("Social Media Hours vs Sleep Hours")
plt.xlabel("Daily Social Media Hours")
plt.ylabel("Sleep Hours")

plt.show()
```

```python
correlation = df.corr(numeric_only=True)

plt.imshow(correlation, cmap='coolwarm')

plt.colorbar()

plt.xticks(range(len(correlation.columns)),
           correlation.columns,
           rotation=90)

plt.yticks(range(len(correlation.columns)),
           correlation.columns)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.fill_between(df.index,
                 df['daily_social_media_hours'],
                 color='lightblue')

plt.title("Area Plot of Social Media Hours")
plt.xlabel("Index")
plt.ylabel("Hours")

plt.show()
```

```python
plt.stem(df['stress_level'].head(20))

plt.title("Stress Level Stem Plot")
plt.xlabel("Index")
plt.ylabel("Stress Level")

plt.show()
```
